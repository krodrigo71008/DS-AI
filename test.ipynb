```python
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
import numpy as np

u = np.arange(0, 1920, 0.5)
v = np.arange(0, 1080, 0.5)

U, V = np.meshgrid(u, v)
x = (21.934635556848825*V - 15.489302596827795*U -1331.6265071679857)/(0.7330518820051866*V+1164.9054339228585)
z = (21.934635556848825*V + 15.489302596827795*U -1331.6265071679857)/(0.7330518820051866*V+1164.9054339228585)

fig = plt.figure()
ax = Axes3D(fig, auto_add_to_figure=False)
fig.add_axes(ax)
ax.plot_surface(U, V, x)
plt.xlabel('u')
plt.ylabel('v')
plt.show()
```

```python
import random

random.seed(727)

std_1 = random.random()*100
std_2 = random.random()*100

cov_mat = np.array([
    [std_1**2, 0],
    [0, std_2**2]
])

np.random.seed(727)

NUM_POINTS = 1000000
points = []
for _ in range(NUM_POINTS):
    points.append([np.random.normal()*std_1, np.random.normal()*std_2])

points = np.array(points)
```

```python
plt.scatter(points[:, 0], points[:, 1], marker='.')
plt.title(f"stds: {std_1:.2f}, {std_2:.2f}")
```

```python
inv_mat = np.linalg.inv(cov_mat)
dists = []
for i in range(points.shape[0]):
    dist = points[i, :] @ inv_mat @ points[i, :].T
    dists.append(dist)
dists = np.array(dists)
```

```python
import pandas as pd
dist_df = pd.Series(dists)
dist_df.describe(percentiles=[0.25, 0.5, 0.75, 0.8, 0.85, 0.9, 0.95, 0.975, 0.99, 0.999])
```
